```python
#selection sort
def selectionSort(l):
    size=len(l)
    
    for ind in range(size):
        min_index = ind
 
        for j in range(ind + 1, size):
            
            if l[j] < l[min_index]:
                min_index = j
        (l[ind], l[min_index]) = (l[min_index], l[ind])
 
list = [2,5,3,7,9,0,2,3,45]
selectionSort(list)
print('The list after sorting in Ascending Order by selection sort is:')
print(list)
```

```python
#bubble sort
def bubblesort(l):
    swapped = False
    for n in range(len(l)-1, 0, -1):
        for i in range(n):
            if l[i] > l[i + 1]:
                swapped = True
                l[i], l[i + 1] = l[i + 1], l[i]       
        if not swapped:
            return
 
list = [54,8,9,5,2,2,9,2,21,4,8,1,0]
 

bubblesort(list)
print("Sorted Array is, ")
print(list)
```

```python
#merge sort

def merge(list, l, m, r):
    n1 = m - l + 1
    n2 = r - m
 
    L = [0] * (n1)
    R = [0] * (n2)
 
    for i in range(0, n1):
        L[i] = list[l + i]
 
    for j in range(0, n2):
        R[j] = list[m + 1 + j]
 
    i = 0    
    j = 0     
    k = l     
 
    while i < n1 and j < n2:
        if L[i] <= R[j]:
            list[k] = L[i]
            i += 1
        else:
            list[k] = R[j]
            j += 1
        k += 1
   
    while i < n1:
        list[k] = L[i]
        i += 1
        k += 1
 
   
    while j < n2:
        list[k] = R[j]
        j += 1
        k += 1
 
def mergeSort(list, l, r):
    if l < r:
 
       
        m = l+(r-l)//2
       
        mergeSort(list, l, m)
        mergeSort(list, m+1, r)
        merge(list, l, m, r)
 
 

list = [1,4,6,4,3,2,4,666,3,2,45,3]

print("Given array is",list)

n=len(list)
mergeSort(list, 0, n-1)
print("Sorted list is",list)

 
```
