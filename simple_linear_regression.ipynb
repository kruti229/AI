```python
from numpy import *
from matplotlib.pyplot import *
import pandas as pd

data=pd.read_csv("C:/Users/kruti/Downloads/studentscores.csv")

print(data.head(10))

```

```python
# now in simple linear regression we only need two para meter
""" linear regression predict the y
Y=mX+c is the equation """
x=data[['Hours']]
y=data[['Scores']]
```

```python
import matplotlib.pyplot as plt
plt.scatter(x,y)
plt.xlabel('Hours')
plt.ylabel('Scores')
```

```python
from sklearn.model_selection import *
X_train,X_test,Y_train,Y_test=train_test_split(x,y,test_size=0.4,random_state=23)
```

```python
#now here our data is one D so we have to convert it into 2D
import numpy as np
```

```python
X_train=np.array(X_train).reshape(-1,1)
X_test=np.array(X_test).reshape(-1,1)
```

```python
print('training data:',X_train)
print('testing data',X_test)

```

```python
# now we have linear regression model from sklearn
from sklearn.linear_model import LinearRegression
lr= LinearRegression()
```

```python
#this will train model
lr.fit(X_train,Y_train)
```

```python
#intercept
c=lr.intercept_
#coefficient for model
m=lr.coef_
```

```python
c
```

```python
y_pred_train=m*X_train+c
y_pred_train
```

```python
y_pred=lr.predict(X_train)
y_pred
```

```python
#now find fit line
plt.scatter(X_train,Y_train)
plt.plot(X_train,y_pred,color='red')
```

```python
#now check with test 
y_pred_test=lr.predict(X_test)
y_pred_test
```

```python
plt.scatter(X_test,Y_test)
plt.plot(X_test,y_pred_test,color='red')
```
