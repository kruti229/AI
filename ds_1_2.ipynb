# Descriptive stastic

```python
# descriptive stastic for iris data
import pandas as pd
import numpy as np
dt=pd.read_csv('C:/Users/kruti/Downloads/iris_data.csv')
print(dt)
```

```python
dt.isnull().sum()
```

```python
dt=dt.drop(columns='Class')
```

```python
print(dt)
```

```python
mn=dt.mean()
print(mn)
```

```python
mdn=dt.median()
print(mdn)
```

```python
md=dt.mode()
print(md)
```

```python
#range in data
sl_range=dt.SepalLength.max()-dt.SepalLength.min()
print('range in SepalLength is:',sl_range)

sw_range=dt.SepalWidth.max()-dt.SepalWidth.min()
print('range in SepalWidth  is:',sw_range)

pl_range=dt.PetalLength.max()-dt.PetalLength.min()
print('range in PetalLength  is:',pl_range)

pw_range=dt.PetalWidth.max()-dt.PetalWidth.min()
print('range in PetalWidth is:',pw_range)
```

```python
#varience
sl_var=dt.SepalLength.var()
print('varience in SepalLength is:',sl_var)

sw_var=dt.SepalWidth.var()
print('varience in SepalWidth is:',sw_var)

pl_var=dt.PetalLength.var()
print('varience in PetalLength is:',pl_var)

pw_var=dt.PetalWidth.var()
print('varience in PetalWidth is:',pw_var)
```

```python
#percentile -for sepalLength
dt_sl=dt.SepalLength 
dt_sl=np.sort(dt_sl)
Q1_sl=np.percentile(dt_sl,25)
Q2_sl=np.percentile(dt_sl,50)
Q3_sl=np.percentile(dt_sl,75)
print('first quartile for SepalLength:',Q1_sl)
print('Second quartile for SepalLength:',Q2_sl)
print('Third quartile for SepalLength:',Q3_sl)
```

```python
#percentile -for sepalWidth
dt_sw=dt.SepalWidth  
dt_sw=np.sort(dt_sw)
Q1_sw=np.percentile(dt_sw,25)
Q2_sw=np.percentile(dt_sw,50)
Q3_sw=np.percentile(dt_sw,75)
print('first quartile for SepalWidth:',Q1_sw)
print('Second quartile for SepalWidth:',Q2_sw)
print('Third quartile for SepalWidth:',Q3_sw)
```

```python
#percentile -for PetalLength
dt_pl=dt.PetalLength  
dt_pl=np.sort(dt_pl)
Q1_pl=np.percentile(dt_pl,25)
Q2_pl=np.percentile(dt_pl,50)
Q3_pl=np.percentile(dt_pl,75)
print('first quartile for PetalLength:',Q1_pl)
print('Second quartile for PetalLength:',Q2_pl)
print('Third quartile for PetalLength:',Q3_pl)
```

```python
#percentile -for PetalWidth
dt_pw=dt.PetalWidth  
dt_pw=np.sort(dt_pw)
Q1_pw=np.percentile(dt_pw,25)
Q2_pw=np.percentile(dt_pw,50)
Q3_pw=np.percentile(dt_pw,75)
print('first quartile for PetalWidth:',Q1_pw)
print('Second quartile for PetalWidth:',Q2_pw)
print('Third quartile for PetalWidth:',Q3_pw)
```

```python
#inter Quartile distance
IQR_sl=Q3_sl-Q1_sl
IQR_sw=Q3_sw-Q1_sw
IQR_pl=Q3_pl-Q1_pl
IQR_pw=Q3_pw-Q1_pw
print('InterQuartile Distance in SepalLength:',IQR_sl)
print('InterQuartile Distance in SepalWidth:',IQR_sw)
print('InterQuartile Distance in PetalLength:',IQR_pl)
print('InterQuartile Distance in PetalWidth:',IQR_pw)
```
