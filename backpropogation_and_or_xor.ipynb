```python
from sklearn.neural_network import MLPClassifier
```

```python
#or gate
x=[[0,0],[0,1],[1,0],[1,1]]
y=[0,1,1,1]
```

```python
clf = MLPClassifier(solver='lbfgs', alpha=1e-5,hidden_layer_sizes=(5, 3), random_state=1)
clf.fit(x,y)
clf.predict([[0.1,0],[1,0],[0.,1],[0.1,0.9],[1,1]])


```

```python
clf.predict([[0.9,0],[1,0.9],[0,1.01],[0.11,0.2],[0.01,1]])
```

```python
clf.predict([[0.12,0.99],[1,0],[0.7,1.02],[0.1,0.03],[0,1]])
```

```python

#and gate
a=[[0,0],[0,1],[1,0],[1,1]]
b=[0,0,0,1]


cla = MLPClassifier(solver='lbfgs', alpha=1e-5,hidden_layer_sizes=(5,4), random_state=1)
cla.fit(a,b)
```

```python
cla.predict([[0.1,0],[1,0],[0.,1],[0.1,0.9],[1,1]])
```

```python
cla.predict([[1,1.1]])
```

```python
#XOR gate
p=[[0,0],[0,1],[1,0],[1,1]]
q=[0,1,1,0]


clx = MLPClassifier(solver='lbfgs', alpha=1e-5,hidden_layer_sizes=(5,4), random_state=1)
clx.fit(p,q)
```

```python
clx.predict([[0.1,0],[1,0],[0.,1],[0.1,0.9],[1,1]])
```

```python
clx.predict([[1,0.9]])
```
